# 7. Tīrīšanas funkcijas: no neapstrādāta teksta uz analizējamu tekstu

Līdz šim darījām daudz ko ar atsevišķām rindām. Tagad šo loģiku saliksim funkcijās, lai kods būtu atkārtoti lietojams un pārskatāms.

```python
import string

def clean_line(line):
    line = line.lower().strip()
    line = line.translate(str.maketrans("", "", string.punctuation))
    line = " ".join(line.split())
    return line

clean_line("  Python, python!!!   ")
```

# 9. Ģeneratori: apstrāde pa vienai rindai

Ģeneratori ir noderīgi, ja negribam visu glabāt atmiņā uzreiz. Tas ir labs tilts uz reālu lielu datu failu apstrādi.

```python
def cleaned_line_generator(lines):
    for line in lines:
        cleaned = clean_line(line)
        if cleaned != "":
            yield cleaned

for item in cleaned_line_generator(sample_lines):
    print(item)
```

# 10. Klases un objekti: vienkārša teksta korpusa klase

Šeit netaisām sarežģītu objektorientētu sistēmu. Mērķis ir saprast pamatideju:
- klase definē struktūru un uzvedību
- objekts glabā konkrētus datus
- metodes palīdz sakārtot darbības ap datiem

```python
class TextCorpus:
    def __init__(self, lines):
        self.raw_lines = lines
        self.cleaned_lines = []
        self.tokens = []

    def clean(self):
        self.cleaned_lines = [clean_line(line) for line in self.raw_lines if clean_line(line) != ""]

    def tokenize(self):
        self.tokens = []
        for line in self.cleaned_lines:
            self.tokens.extend(line.split())

    def word_counts(self):
        counts = {}
        for token in self.tokens:
            counts[token] = counts.get(token, 0) + 1
        return counts

corpus = TextCorpus(sample_lines)
corpus.clean()
corpus.tokenize()
corpus.word_counts()
```
